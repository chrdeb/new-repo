**5- What is the highest amount of BasePay in the dataset ?**

**6- How many Employees do not get Benefits? (How many NaN records do we have in the Benefits column?)**

```python
no_benefits_count = sal['Benefits'].isnull().sum()
print(no_benefits_count)
```

**10- What is the name of highest paid person (including benefits)?**

```python
highest_paid_person = sal.loc[sal['TotalPayBenefits'].idxmax()]['EmployeeName']
print(highest_paid_person)

# I used .idxmax() to get the index of the highest value in the TotalPayBenefits column. 
# I used .loc to retrieve the row corresponding to that index
# From that row I pulled the 'EmployeeName'
```

**11- What was the average (mean) TotalPayBenefits for all employees per year? (2011-2014) ?**

```python
yearly_average = sal.groupby('Year')['TotalPayBenefits'].mean()
print(yearly_average)
```

**12- How many unique EmployeeName are there in the dataset?**

**13- What are the top 5 most common jobs?**

```python
most_common_jobs = sal['JobTitle'].value_counts().head()
print(most_common_jobs)

# value_counts() to count the frequency of unique values in 'JobTitle' column
# head() to display the top 5 unique values
```

**14- How many Job Titles were represented by only three persons in 2011? (e.g. Job Titles with only three occurences in 2011?)**

```python
sal_counts_2011 = sal[sal['Year'] == 2011]['JobTitle'].value_counts()
three_person_count = sum(sal_counts_2011 == 3)
print(three_person_count)
```

**15- How many people have the word Police in their job title?**

```python
police_count = filter(lambda x: 'Police' in x, sal['JobTitle'])
print(len(list(police_count)))

# lambda acts as a for loop iterating on the JobTitle column and filtering for when 'Police' is in x
# in order to use len() on police_count, must make it a list(), now we can count the number of people with 'Police' in their job title
```

**18- Is there a correlation between length of the Job Title string and Salary?**

```python
sal['title_len'] = sal['JobTitle'].apply(len)
correlation_matrix = np.corrcoef(sal['title_len'], sal['TotalPayBenefits'])
correlation_value = correlation_matrix[0, 1]
print(correlation_value)
```
